```python
import pandas as pd 
s= pd.Series([10,None,20,None,30,40,None])
print(s.isnull())
```

```python
import pandas as pd

s= pd.Series([10,None,20,None,30,40,None])
print(s.isnull())
print(s.notnull())
print(s.fillna("NA"))
print(s.dropna())
print(s.isna())
```

```python
import pandas as pd
s= pd.Series([10,None,20,None,30,40,None])
print(s.fillna(s.mean()))
print(s.fillna(s.ffill()))
print(s.fillna(s.bfill()))
```
