## Section 2: Missing-Value Methods

```python
#15. Write a Python program to create a Series containing 10, None, 30, None, 50 and display it.
s= pd.Series([10,None,30,None,50])
print(s)
```

```python
#16. Write a Python program to check missing values in a Series using isnull().
s= pd.Series([10,None,30,None,50])
print(s.isnull())
```

```python
#17. Write a Python program to check non-missing values in a Series using notnull().
s= pd.Series([10,None,30,None,50])
print(s.notnull())
```

```python
#18. Write a Python program to check missing values in a Series using isna()
s= pd.Series([10,None,30,None,50])
print(s.isna())
```

```python
#19. Write a Python program to replace all missing values in a Series with 0 using fillna().
s= pd.Series([10,None,30,None,50])
print(s.fillna(0))
```

```python
#20. Write a Python program to remove all missing values from a Series using dropna().
s= pd.Series([10,None,30,None,50])
print(s.dropna())
```

```python
#21. Write a Python program to replace missing values with the mean of the Series using fillna().
s= pd.Series([10,None,30,None,50])
print(s.fillna(s.mean()))
```

```python
#22. Write a Python program to replace missing values with the previous valid value using ffill().
s= pd.Series([10,None,30,None,50])
print(s.fillna(s.ffill()))
```

```python
#23. Write a Python program to replace missing values with the next valid value using bfill().
s= pd.Series([10,None,30,None,50])
print(s.fillna(s.bfill()))
```
